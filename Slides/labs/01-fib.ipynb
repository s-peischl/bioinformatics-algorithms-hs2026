## Naive recursion

Each call splits into two. The recursion tree for $F_5$ has repeated $F_3$s.


```python
def fib(n):
    # Follows the definition. Correct, and exponential.
    if n < 2:
        return n
    return fib(n - 2) + fib(n - 1)


print("fib(5) =", fib(5))
print("fib(10) =", fib(10))
```

## Store previous values

Same recurrence, one pass. Change `n` below if you want.


```python
def fib1(n):
    # F[i] is F_i. We never recompute a value.
    F = [0, 1]
    for i in range(2, n + 1):
        F.append(F[i - 1] + F[i - 2])
    return F[n]


print("fib1(5) =", fib1(5))
print("fib1(10) =", fib1(10))
# print("fib1(30) =", fib1(30))
```

## Time both

Do not start at $n=40$ for `fib`. 25 is already enough to feel the gap.


```python
import time

for n in (5, 10, 15, 20, 25):
    t0 = time.perf_counter(); fib(n); rec = time.perf_counter() - t0
    t0 = time.perf_counter(); fib1(n); lin = time.perf_counter() - t0
    print(f"n={n:2d}   fib {rec:.4f}s   fib1 {lin:.6f}s")
```

## Factorial: recursion that is *not* a tree

$n! = 1 \cdot 2 \cdots n$, or $n \cdot (n-1)!$ with $0! = 1$. `fact` makes **one** recursive call. `fib` made two.


```python
def fact(n):
    # n * (n-1) * ... * 1. One recursive call.
    if n <= 1:
        return 1
    return n * fact(n - 1)


def fact1(n):
    # Same product, left to right.
    p = 1
    for k in range(1, n + 1):
        p *= k
    return p


print("fact(5)  =", fact(5), "  (should be 120)")
print("fact1(5) =", fact1(5))
# print("fact(20) =", fact(20))
# Compare the call: fact(n) waits on fact(n-1) only. fib(n) waits on fib(n-1) and fib(n-2).
```

## Output length: Hanoi

The algorithm is three lines. The printout is $2^n-1$ lines. `hanoi(3)` is 7 lines you can check. Do not run $n=20$ — about a million lines. Models can *state* the recurrence and still fail to *emit* the transcript (token limits).


```python
def hanoi(n, src, dst, aux):
    # Tiny algorithm. The cost you feel is how many lines it prints.
    if n == 0:
        return
    hanoi(n - 1, src, aux, dst)
    print(src, "->", dst)
    hanoi(n - 1, aux, dst, src)


hanoi(3, "start", "goal", "aux")
# n=4 is 15 lines. n=20 is about a million — do not print it.
```
